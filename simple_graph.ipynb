```python
from typing import TypedDict

from langgraph.constants import START 


class portfoliostate(TypedDict):
    amount_usd : float
    total_usd : float
    total_inr : float
```

```python
def cacl_total(state: portfoliostate) -> portfoliostate:
    state['total_usd'] = state['amount_usd'] * 1.08
    return state

def convert_to_inr(state: portfoliostate) -> portfoliostate:
    state['total_inr'] = state['amount_usd'] * 92
    return state
```

from

```python
from langgraph.graph import StateGraph, START ,END


builder = StateGraph(portfoliostate)

builder.add_node("cal total", cacl_total)

builder.add_node("convert_to_inr", convert_to_inr)

builder.add_edge(START , "cal total")
builder.add_edge( "cal total", "convert_to_inr")
builder.add_edge("convert_to_inr" ,END)

```

```python
graph = builder.compile()
```

```python
from IPython.display import Image , display



display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
graph.invoke({"amount_usd": 100})
```
